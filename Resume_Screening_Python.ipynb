<h1>Cleaning Data: </h1> <br> 1.URLs <br> 2.hashtags <br> 3.mentions <br>4.special letters <br> 5. punctuations

```python
# Regular Expression
import re
def cleanResume(txt):
    cleanText = re.sub('http\S+\s', ' ', txt)
    cleanText = re.sub('RT|cc', ' ', cleanText)
    cleanText = re.sub('#\S+\s', ' ', cleanText)
    cleanText = re.sub('@\S+', '  ', cleanText)  
    cleanText = re.sub('[%s]' % re.escape("""!"#$%&'()*+,-./:;<=>?@[\]^_`{|}~"""), ' ', cleanText)
    cleanText = re.sub(r'[^\x00-\x7f]', ' ', cleanText) 
    cleanText = re.sub('\s+', ' ', cleanText)
    return cleanText
```

```python
df['Resume'] = df['Resume'].apply(lambda x: cleanResume(x))
```

<h1> Words into Categorical Values</h1>

```python
from sklearn.preprocessing import LabelEncoder 
le = LabelEncoder()
```

```python
le.fit(df['Category']) 
df['Category'] = le.transform(df['Category'])
```

```python
print(df['Category'])
```

 <h1>Vectorization</h1>

```python
# term frequency inverse document vectorization
from sklearn.feature_extraction.text import TfidfVectorizer 
tfidf = TfidfVectorizer(stop_words='english')


tfidf.fit(df['Resume']) 
requiredTaxt = tfidf.transform(df['Resume'])
```

```python
df
```

<h1> Splitting</h1>

```python
from sklearn.model_selection import train_test_split 
```

```python
# X is input, 
X_train, X_test, y_train, y_test = train_test_split(requiredTaxt, df['Category'], test_size=0.2, random_state=42)

```

<h1>Now let's train the model and print the classification report: </h1>

```python
# Since this is a classification problem, we will implement the KNeighborsClassifier
from sklearn.neighbors import KNeighborsClassifier 
from sklearn.multiclass import OneVsRestClassifier # help in better prediction
from sklearn.metrics import accuracy_score

# bagging, boosting and ensembling

clf = OneVsRestClassifier(KNeighborsClassifier()) 
clf.fit(X_train, y_train) 
ypred = clf.predict(X_test)

print(ypred)
```
